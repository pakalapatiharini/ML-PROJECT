```python
# ============================================================
# FULL EDA - HEART DISEASE DATASET
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

CSV_PATH = "heart_noisy_for_eda.csv"
df = pd.read_csv(CSV_PATH)

print("=" * 70)
print("HEART DISEASE DATASET - EXPLORATORY DATA ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 2. FIRST 5 ROWS
# ------------------------------------------------------------

print("\n1. FIRST 5 ROWS")
print(df.head())


# ------------------------------------------------------------
# 3. LAST 5 ROWS
# ------------------------------------------------------------

print("\n2. LAST 5 ROWS")
print(df.tail())


# ------------------------------------------------------------
# 4. RANDOM 5 ROWS
# ------------------------------------------------------------

print("\n3. RANDOM 5 ROWS")
print(df.sample(5))


# ------------------------------------------------------------
# 5. DATASET SHAPE
# ------------------------------------------------------------

print("\n4. DATASET SHAPE")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])


# ------------------------------------------------------------
# 6. COLUMN NAMES
# ------------------------------------------------------------

print("\n5. COLUMN NAMES")
print(df.columns.tolist())


# ------------------------------------------------------------
# 7. DATA TYPES
# ------------------------------------------------------------

print("\n6. DATA TYPES")
print(df.dtypes)


# ------------------------------------------------------------
# 8. DATASET INFORMATION
# ------------------------------------------------------------

print("\n7. DATASET INFORMATION")
df.info()


# ------------------------------------------------------------
# 9. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\n8. STATISTICAL SUMMARY")
print(df.describe())


# ------------------------------------------------------------
# 10. MISSING VALUES
# ------------------------------------------------------------

print("\n9. MISSING VALUES")
print(df.isnull().sum())


# ------------------------------------------------------------
# 11. MISSING VALUE PERCENTAGE
# ------------------------------------------------------------

print("\n10. MISSING VALUE PERCENTAGE")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)


# ------------------------------------------------------------
# 12. DUPLICATE VALUES
# ------------------------------------------------------------

print("\n11. DUPLICATE ROWS")
print("Number of duplicate rows:", df.duplicated().sum())


# ------------------------------------------------------------
# 13. UNIQUE VALUES
# ------------------------------------------------------------

print("\n12. UNIQUE VALUES")
print(df.nunique())


# ------------------------------------------------------------
# 14. TARGET DISTRIBUTION
# ------------------------------------------------------------

print("\n13. TARGET DISTRIBUTION")
print(df["target"].value_counts())


# ------------------------------------------------------------
# 15. TARGET PERCENTAGE
# ------------------------------------------------------------

print("\n14. TARGET PERCENTAGE")
print(df["target"].value_counts(normalize=True) * 100)


# ------------------------------------------------------------
# 16. NUMERICAL COLUMNS
# ------------------------------------------------------------

numeric_cols = df.select_dtypes(include=np.number).columns

print("\n15. NUMERICAL COLUMNS")
print(numeric_cols.tolist())


# ------------------------------------------------------------
# 17. CATEGORICAL COLUMNS
# ------------------------------------------------------------

categorical_cols = df.select_dtypes(
    exclude=np.number
).columns

print("\n16. CATEGORICAL COLUMNS")
print(categorical_cols.tolist())


# ------------------------------------------------------------
# 18. TARGET COUNT PLOT
# ------------------------------------------------------------

plt.figure(figsize=(6, 4))
sns.countplot(x="target", data=df)
plt.title("Heart Disease Target Distribution")
plt.xlabel("Target")
plt.ylabel("Number of Patients")
plt.show()


# ------------------------------------------------------------
# 19. NUMERICAL FEATURE DISTRIBUTIONS
# ------------------------------------------------------------

df[numeric_cols].hist(
    figsize=(16, 12),
    bins=20
)

plt.suptitle("Distribution of Numerical Features")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 20. AGE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 4))
sns.histplot(df["age"], bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()


# ------------------------------------------------------------
# 21. SEX DISTRIBUTION
# ------------------------------------------------------------

print("\n17. SEX DISTRIBUTION")
print(df["sex"].value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(x="sex", data=df)
plt.title("Sex Distribution")
plt.xlabel("Sex")
plt.ylabel("Number of Patients")
plt.show()


# ------------------------------------------------------------
# 22. CHEST PAIN DISTRIBUTION
# ------------------------------------------------------------

print("\n18. CHEST PAIN DISTRIBUTION")
print(df["cp"].value_counts())

plt.figure(figsize=(7, 4))
sns.countplot(x="cp", data=df)
plt.title("Chest Pain Type Distribution")
plt.xlabel("Chest Pain Type")
plt.ylabel("Number of Patients")
plt.show()


# ------------------------------------------------------------
# 23. RESTING BLOOD PRESSURE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 4))
sns.histplot(df["trestbps"], bins=20, kde=True)
plt.title("Resting Blood Pressure Distribution")
plt.xlabel("Resting Blood Pressure")
plt.ylabel("Frequency")
plt.show()


# ------------------------------------------------------------
# 24. CHOLESTEROL DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 4))
sns.histplot(df["chol"], bins=20, kde=True)
plt.title("Cholesterol Distribution")
plt.xlabel("Cholesterol")
plt.ylabel("Frequency")
plt.show()


# ------------------------------------------------------------
# 25. MAXIMUM HEART RATE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 4))
sns.histplot(df["thalach"], bins=20, kde=True)
plt.title("Maximum Heart Rate Distribution")
plt.xlabel("Maximum Heart Rate")
plt.ylabel("Frequency")
plt.show()


# ------------------------------------------------------------
# 26. HEART DISEASE BY SEX
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.countplot(x="sex", hue="target", data=df)
plt.title("Heart Disease by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Patients")
plt.show()


# ------------------------------------------------------------
# 27. HEART DISEASE BY CHEST PAIN
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.countplot(x="cp", hue="target", data=df)
plt.title("Heart Disease by Chest Pain Type")
plt.xlabel("Chest Pain Type")
plt.ylabel("Number of Patients")
plt.show()


# ------------------------------------------------------------
# 28. HEART DISEASE BY EXERCISE-INDUCED ANGINA
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.countplot(x="exang", hue="target", data=df)
plt.title("Exercise-Induced Angina vs Heart Disease")
plt.xlabel("Exercise-Induced Angina")
plt.ylabel("Number of Patients")
plt.show()


# ------------------------------------------------------------
# 29. AGE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.boxplot(x="target", y="age", data=df)
plt.title("Age vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Age")
plt.show()


# ------------------------------------------------------------
# 30. CHOLESTEROL VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.boxplot(x="target", y="chol", data=df)
plt.title("Cholesterol vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Cholesterol")
plt.show()


# ------------------------------------------------------------
# 31. BLOOD PRESSURE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.boxplot(x="target", y="trestbps", data=df)
plt.title("Resting Blood Pressure vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Resting Blood Pressure")
plt.show()


# ------------------------------------------------------------
# 32. MAXIMUM HEART RATE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.boxplot(x="target", y="thalach", data=df)
plt.title("Maximum Heart Rate vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Maximum Heart Rate")
plt.show()


# ------------------------------------------------------------
# 33. OLDPEAK VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
sns.boxplot(x="target", y="oldpeak", data=df)
plt.title("Oldpeak vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Oldpeak")
plt.show()


# ------------------------------------------------------------
# 34. CORRELATION MATRIX
# ------------------------------------------------------------

correlation = df.corr(numeric_only=True)

print("\n19. CORRELATION MATRIX")
print(correlation)


# ------------------------------------------------------------
# 35. CORRELATION HEATMAP
# ------------------------------------------------------------

plt.figure(figsize=(12, 9))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()


# ------------------------------------------------------------
# 36. CORRELATION WITH TARGET
# ------------------------------------------------------------

print("\n20. CORRELATION WITH TARGET")
print(
    correlation["target"]
    .sort_values(ascending=False)
)


# ------------------------------------------------------------
# 37. OUTLIER DETECTION USING IQR
# ------------------------------------------------------------

print("\n21. OUTLIER ANALYSIS")

for column in numeric_cols:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(column, ":", len(outliers), "outliers")


# ------------------------------------------------------------
# 38. BOXPLOTS FOR ALL NUMERICAL FEATURES
# ------------------------------------------------------------

plt.figure(figsize=(15, 8))

df[numeric_cols].boxplot()

plt.title("Boxplot of Numerical Features")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 39. SKEWNESS
# ------------------------------------------------------------

print("\n22. SKEWNESS")
print(df[numeric_cols].skew())


# ------------------------------------------------------------
# 40. KURTOSIS
# ------------------------------------------------------------

print("\n23. KURTOSIS")
print(df[numeric_cols].kurtosis())


# ------------------------------------------------------------
# 41. GROUP ANALYSIS BY TARGET
# ------------------------------------------------------------

print("\n24. AVERAGE VALUES BY TARGET")

print(
    df.groupby("target")[numeric_cols]
    .mean()
)


# ------------------------------------------------------------
# 42. MEDIAN VALUES BY TARGET
# ------------------------------------------------------------

print("\n25. MEDIAN VALUES BY TARGET")

print(
    df.groupby("target")[numeric_cols]
    .median()
)


# ------------------------------------------------------------
# 43. FINAL DATASET SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL EDA SUMMARY")
print("=" * 70)

print("Total Records        :", df.shape[0])
print("Total Columns        :", df.shape[1])
print("Total Features       :", df.shape[1] - 1)
print("Target Variable      : target")
print("Missing Values       :", df.isnull().sum().sum())
print("Duplicate Rows       :", df.duplicated().sum())
print("Target 0             :", (df["target"] == 0).sum())
print("Target 1             :", (df["target"] == 1).sum())
print("=" * 70)
```
